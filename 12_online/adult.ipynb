# Adult dataset

- `education-num`: continuous.
- `marital-status`: Married-civ-spouse, Divorced, Never-married, Separated, Widowed, Married-spouse-absent, Married-AF-spouse.
- `occupation`: Tech-support, Craft-repair, Other-service, Sales, Exec-managerial, Prof-specialty, Handlers-cleaners, Machine-op-inspct, Adm-clerical, Farming-fishing, Transport-moving, Priv-house-serv, Protective-serv, Armed-Forces.

- `sex`: Female, Male.
- `capital-gain`: continuous.
- `capital-loss`: continuous.
- `hours-per-week`: continuous.
- `native-country`: United-States, Cambodia, England, Puerto-Rico, Canada, Germany, Outlying-US(Guam-USVI-etc), India, Japan, Greece, South, China, Cuba, Iran, Honduras, Philippines, Italy, Poland, Jamaica, Vietnam, Mexico, Portugal, Ireland, France, Dominican-Republic, Laos, Ecuador, Taiwan, Haiti, Columbia, Hungary, Guatemala, Nicaragua, Scotland, Thailand, Yugoslavia, El-Salvador, Trinadad&Tobago, Peru, Hong, Holand-Netherlands.

### 1. (2đ) Import các thư viện cần thiết

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

### 2. (2đ) Đọc dữ liệu từ file này  [address](https://github.com/thieu1995/csv-files/blob/main/data/pandas/adult.data) và gán tên biến là df. Đồng thời thiết lập tên các features dựa trên thông tin đã cho ở đầu bài.

```python
url = "https://raw.githubusercontent.com/thieu1995/csv-files/main/data/pandas/adult.data"
column_names = [
    'age', 'workclass', 'fnlwgt', 'education', 'education-num',
    'marital-status', 'occupation', 'relationship', 'race', 'sex',
    'capital-gain', 'capital-loss', 'hours-per-week', 'native-country', 'salary'
]

df = pd.read_csv(url, names=column_names, sep=',\\s*', engine='python')
df.head()
```

### 7. (3đ) Tạo pie chart thể hiện phần trăm của cột race và sex (sẽ có 5*2=10 categories tổng thể trong 1 hình vẽ). Nhận xét xem nhóm nào chiếm tỉ lệ nhiều nhất.

```python
# Tạo cột mới kết hợp race và sex
df['race_sex'] = df['race'] + ' - ' + df['sex']

# Tính toán số lượng cho từng category
race_sex_counts = df['race_sex'].value_counts()

# Vẽ biểu đồ pie
plt.figure(figsize=(12, 10))
plt.pie(race_sex_counts, labels=race_sex_counts.index, autopct='%1.1f%%', startangle=140, colors=sns.color_palette('tab10'))
plt.title('Percentage Distribution of Race and Sex Categories')
plt.axis('equal')
plt.show()

# Nhận xét
most_common = race_sex_counts.idxmax()
percentage = (race_sex_counts.max() / len(df)) * 100
print(f"Nhận xét: Nhóm chiếm tỉ lệ nhiều nhất là '{most_common}' với khoảng {percentage:.1f}%.")
```

### 18. (6đ) Có đúng là những người kiếm được hơn 50K đều có trình độ học vấn ít nhất là  high school? (education feature – Bachelors, Prof-school, Assoc-acdm, Assoc-voc, Masters or Doctorate)

```python
high_income = df[df['salary'] == '>50K']
edu_list = high_income['education'].unique()
print(f"Education levels of >50K earners: {edu_list}")

# Checking for Bachelor's or higher as defined in the prompt
advanced_degrees = ['Bachelors', 'Prof-school', 'Assoc-acdm', 'Assoc-voc', 'Masters', 'Doctorate']
missing_degrees = high_income[~high_income['education'].isin(advanced_degrees)]

if missing_degrees.empty:
    print("\nYes, all individuals earning >50K have the specified education levels.")
else:
    print(f"\nNo, there are {len(missing_degrees)} individuals earning >50K with other education levels like {missing_degrees['education'].unique()}.")
```

### 20. (6đ) Tỉ lệ người married hay single men (marital-status feature) chiếm nhiều hơn trong số những người kiếm được >50k? Những người có tình trạng hôn nhân (married) có marital-status feature bắt đầu với Married (Married-civ-spouse, Married-spouse-absent or Married-AF-spouse), còn lại được coi là single men.

```python
high_income_men = df[(df['salary'] == '>50K') & (df['sex'] == 'Male')]

# Define married as statuses starting with 'Married'
married_statuses = ['Married-civ-spouse', 'Married-spouse-absent', 'Married-AF-spouse']

married_men_count = high_income_men[high_income_men['marital-status'].isin(married_statuses)].shape[0]
single_men_count = high_income_men[~high_income_men['marital-status'].isin(married_statuses)].shape[0]

print(f'Among men earning >50K:')
print(f'- Married men: {married_men_count}')
print(f'- Single men: {single_men_count}')

if married_men_count > single_men_count:
    print('\nNhᤩn xét: Nhóm married men chiếm tỷ lệ nhiều hơn.')
else:
    print('\nNhᤩn xét: Nhóm single men chiếm tỷ lệ nhiều hơn.')
```

### 21. (6đ) Số giờ tối đa một người làm việc mỗi tuần (hours-per-week feature) là bao nhiêu? Có bao nhiêu người làm việc với số giờ như vậy và tỷ lệ những người kiếm được nhiều (>50K) trong số đó là bao nhiêu?

```python
max_hours = df['hours-per-week'].max()
num_max_workers = len(df[df['hours-per-week'] == max_hours])

high_income_max_workers = df[(df['hours-per-week'] == max_hours) & (df['salary'] == '>50K')]
percentage_high_income = (len(high_income_max_workers) / num_max_workers) * 100

print(f"Maximum working hours per week: {max_hours}")
print(f"Number of people working {max_hours} hours/week: {num_max_workers}")
print(f"Percentage of these people earning >50K: {percentage_high_income:.2f}%")
```

### 22. (10đ) Tính thời gian làm việc trung bình (số giờ mỗi tuần) của những người kiếm được ít và nhiều (salary feature) cho mỗi quốc gia (native-country feature). Với đất nước Nhật Bản thì kết quả như nào?

```python
avg_hours_country = df.groupby(['native-country', 'salary'])['hours-per-week'].mean()
print("Average working hours per week by country and salary:")
print(avg_hours_country)

if 'Japan' in df['native-country'].values:
    japan_stats = avg_hours_country.loc['Japan']
    print("\nAverage working hours in Japan:")
    print(japan_stats)
else:
    print("\nJapan not found in the dataset.")
```

### 23. (10đ) Số người da trắng (white in race feature) kiếm được số tiền >50k ở nước Mỹ chiếm bao nhiêu phần trăm. Câu hỏi tương tự với người da đen (black in race feature).

```python
us_data = df[df['native-country'] == 'United-States']

# White people in US earning >50K
white_us = us_data[us_data['race'] == 'White']
white_high_income = len(white_us[white_us['salary'] == '>50K'])
white_percentage = (white_high_income / len(white_us)) * 100

# Black people in US earning >50K
black_us = us_data[us_data['race'] == 'Black']
black_high_income = len(black_us[black_us['salary'] == '>50K'])
black_percentage = (black_high_income / len(black_us)) * 100

print(f"Percentage of White people in the US earning >50K: {white_percentage:.2f}%")
print(f"Percentage of Black people in the US earning >50K: {black_percentage:.2f}%")
```
